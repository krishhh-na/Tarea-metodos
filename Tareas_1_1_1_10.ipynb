```python
import matplotlib.pyplot as plt
import numpy as np
import math
x = np.linspace(-8,8,400)
y = x * np.sin(x)
taylor = 0
for n in range(10):
  taylor = taylor + (-1) ** n * x **(2*n + 1) / math.factorial(2*n +1)
plt.plot(x, y, label = 'x sin(x)', linestyle='-', color = 'pink', linewidth= 2)
plt.plot(x, taylor, label = 'Taylor', linestyle='--', color='teal', linewidth=2)
plt.legend()
plt.grid()
plt.show()


```

```python
%matplotlib inline
import h5py
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. CREAR EL ARCHIVO DE PRUEBA .hd5
# ==========================================
nombre_archivo = "datos_prueba.hd5"

# Generamos 10 puntos de datos simulados
x_datos = np.linspace(0, 10, 10)
y_datos = np.sin(x_datos) + 2
e_datos = np.random.uniform(0.1, 0.4, 10)  # Errores aleatorios

# Guardamos los datasets en el archivo HDF5
with h5py.File(nombre_archivo, 'w') as f:
    f.create_dataset('x', data=x_datos)
    f.create_dataset('y', data=y_datos)
    f.create_dataset('e', data=e_datos)

print(f"¡Archivo '{nombre_archivo}' creado exitosamente!")

# ==========================================
# 2. TU FUNCIÓN PARA LA TAREA
# ==========================================
def graficar_datos_h5(ruta_archivo):
    with h5py.File(ruta_archivo, 'r') as f:
        x = f['x'][:]
        y = f['y'][:]
        e = f['e'][:]

    plt.figure(figsize=(8, 5))
    plt.errorbar(
        x, y,
        yerr=e,
        fmt='o',
        capsize=4,
        ecolor='red',
        color='blue',
        label='Datos con error'
    )
    plt.xlabel('Eje X')
    plt.ylabel('Eje Y')
    plt.title('Gráfico de Datos con Barras de Error')
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend()
    plt.show()

# ==========================================
# 3. PROBAR LA FUNCIÓN
# ==========================================
graficar_datos_h5(nombre_archivo)
```
