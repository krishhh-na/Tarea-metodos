```python
import math                                     # importamos librería
suma = 0                                        # suma = cero, para que el conteo empieze 'limpio'
n = 0                                           # n toma el valor 0
while abs(math.exp(1) - suma) >=0.01:           # mide el error entre el valor real y la aproximación mientras es mayor a 0.01
    suma += 1/ math.factorial(n)                # la sumatoria
    n += 1                                      # hace que avanze al siguiente término
print ("e aprox =", suma)                       # mostramos el valor de Euler
print("error=", abs(math.exp(1) - suma))        # mostramos el valor del error
print("terminos usados=", n)                    # mostramos cuantas veces se repitió el bucle
```

```python
import numpy as np
import h5py

# --- 1. Crear el archivo simulacion.txt de ejemplo ---
texto_ejemplo = """3.141590000000000094e-03 2.557680595776000416e+02 4.501400409434224770e+01
6.283180000000000189e-03 3.752818426478838205e+02 2.947770471609068110e+01
9.424770000000000716e-03 4.246697882492646841e+02 2.843524433400840579e+01
1.256636000000000038e-02 4.676325636948347437e+02 2.576223788106328882e+01
1.570795000000000177e-02 4.820912691815412359e+02 1.530261598082400276e+01"""

with open("simulacion.txt", "w") as f:
    f.write(texto_ejemplo)

# --- 2. Función solicitada ---
def txt_a_hd5(archivo_txt, archivo_hd5):
    col1, col2, col3 = np.loadtxt(archivo_txt, unpack=True)

    with h5py.File(archivo_hd5, 'w') as f:
        f.create_dataset('x', data=col1)
        f.create_dataset('y', data=col2)
        f.create_dataset('e', data=col3)

# --- 3. Ejecutar y verificar ---
txt_a_hd5("simulacion.txt", "simulacion.hd5")

# Verificamos los datos guardados
with h5py.File("simulacion.hd5", "r") as f:
    print("Datasets creados:", list(f.keys()))
    print("Valores de 'x':", f['x'][:])
```

```python
import h5py

def txt_a_hd5(nombre_txt, nombre_hd5):

    propiedad1 = []
    propiedad2 = []
    propiedad3 = []

    with open(nombre_txt, 'r') as file:
        for linea in file:

            dato1, dato2, dato3 = linea.split()

            propiedad1.append(float(dato1))
            propiedad2.append(float(dato2))
            propiedad3.append(float(dato3))

    with h5py.File(nombre_hd5, 'w') as file:

        file.create_dataset('propiedad1', data=propiedad1)
        file.create_dataset('propiedad2', data=propiedad2)
        file.create_dataset('propiedad3', data=propiedad3)

txt_a_hd5('simulacion.txt', 'simulacion.hd5')
```

```python
import matplotlib.pyplot as plt                                                     # importamos librerías
import numpy as np
import math
x = np.linspace(-8,8,400)                                                           # crea una lista de 400 números decimales entre el -8 y 8
y = x * np.sin(x)                                                                   # escribimos las funciones
taylor = 0                                                                          # taylor = cero, inicia el acumulador de la serie
for n in range(10):                                                                 # definimos que n va de 0 hasta 9
  taylor = taylor + (-1) ** n * x **(2*n + 2) / math.factorial(2*n +1)              # (-1) ** n hace que se alterne entre positivo y negativo en cada vuelta
                                                                                    #  x **(2*n + 2) eleva los 400 puntos a potencias impares
                                                                                    #  math.factorial(2*n +1) divide  cada punto por el factorial impar
                                                                                    # taylor + suma ese resultado a lo que ya teniamos
plt.plot(x, y, label = 'x sin(x)', linestyle='-', color = 'pink', linewidth= 2)     # definimos los datos, la etiqueta, tipo de linea, color
plt.plot(x, taylor, label = 'Taylor', linestyle='--', color='teal', linewidth=2)
plt.legend()                                                                        # para identificar que significa las curvas
plt.ylim(-8,8)
plt.grid()
plt.show()


```

Suponga que usted es un ayudante de un laboratorio de física. En un experimento se pide medir la posición de una masa que cae en un plano inclinado a intervalos regulares en el tiempo. El equipo disponible puede medir las posiciones con una precisión de
. Uno de los estudiantes usa los siguientes datos (en
):

datos = [0.9,  2. ,  3.6,  5.6,  8. , 10.9, 14.2, 18.]

Grafique estos datos con sus barras de error. Explique por qué el estudiante merece una nota de 1.0. Compare con los siguientes datos obtenidos por un estudiante que merece 7.0.

datos_7 = [0.3,  1.7,  4.1,  5.7,  8.5, 10.7, 13.6, 17.2]

```python
import numpy as np                                                           # importamos librerías
import matplotlib.pyplot as plt

datos_1 = np.array([0.9, 2.0, 3.6, 5.6, 8.0, 10.9, 14.2, 18.0])              # definimos datos
datos_7 = np.array([0.3, 1.7, 4.1, 5.7, 8.5, 10.7, 13.6, 17.2])
error = 0.4

t = np.arange(len(datos_1))                                                  # le asignamos un tiempo a cada medición

fig = plt.figure()                                                           # gernamos una 'figura' para que los gráficos queden dentro de ella
ax1 = fig.add_subplot(1,2,1)                                                 # dividimos la figura en dos gráficos
ax2 = fig.add_subplot(1,2,2)
ax1.errorbar (t, datos_1, yerr = error, fmt = 'o-', color = 'blue',capsize=5, elinewidth=3, capthick=2) # definimos las características para la barras de errores
ax2.errorbar (t, datos_7, yerr = error, fmt = 'o-', color = 'orange',capsize=5, elinewidth=3, capthick=2)

fig.suptitle('Comparación')         # definimos títulos
ax1.set_title('Alumno nota 1.0')
ax2.set_title('Alumno nota 7.0')
ax1.set_xlabel('Tiempo(s)')         # definimos títulos de ejes
ax1.set_ylabel('Posición(cm)')
ax2.set_xlabel('Tiempo(s)')
ax2.set_ylabel('Posición(cm)')
fig.tight_layout(pad=2.0)          # usamos 'pad' para que los bordes y subgraficos no se interpongan entre si
fig.show()
```

El alumno merece nota 1.0 porque la variación entre sus datos es muy perfecta para ser mediciones de un experimento, además, en su gráfico la parábola es casi perfecta, lo que nos quiere decir que inventó o arregló sus datos, para que estos coincidieran con la presición de 4 mm.

```python
%matplotlib inline
import h5py
import numpy as np
import matplotlib.pyplot as plt

nombre_archivo = "datos_prueba.hd5"

x_datos = np.linspace(0, 10, 10)
y_datos = np.sin(x_datos) + 2
e_datos = np.random.uniform(0.1, 0.4, 10)

with h5py.File(nombre_archivo, 'w') as f:
    f.create_dataset('x', data=x_datos)
    f.create_dataset('y', data=y_datos)
    f.create_dataset('e', data=e_datos)

def graficar_datos_h5(ruta_archivo):
    with h5py.File(ruta_archivo, 'r') as f:
        x = f['x'][:]
        y = f['y'][:]
        e = f['e'][:]

    plt.figure(figsize=(8, 5))
    plt.errorbar(
        x, y,
        yerr=e,
        fmt='o',
        capsize=4,
        ecolor='red',
        color='blue',
        label='Datos con error'
    )
    plt.xlabel('Eje X')
    plt.ylabel('Eje Y')
    plt.title('Gráfico de Datos con Barras de Error')
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend()
    plt.show()


graficar_datos_h5(nombre_archivo)
```
